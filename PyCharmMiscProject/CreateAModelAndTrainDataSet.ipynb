# **My First ML Project**

```python
import pandas as pd
df = pd.read_csv('https://raw.githubusercontent.com/dataprofessor/data/refs/heads/master/delaney_solubility_with_descriptors.csv')
df
```

## Data preparation

### Data separation as X and Y

```python
y= df['logS']
y
```

```python
X= df.drop('logS',axis=1) # axis 1 will enable us to drop a column
X
```

### Data separation as a training and test set

```python
from sklearn.model_selection import train_test_split
```

```python
X_train, X_test,y_train,y_test = train_test_split(X,y,test_size=0.2,random_state=100)
```

```python
X_train
```

```python
y_train #80 percent of the data
```

```python
X_test #20 percent of the data
```

## Model Building

### Linear Regression

#### Training a model

```python
from sklearn.linear_model import LinearRegression
lr = LinearRegression()
lr.fit(X_train,y_train)
```

#### Applying the model to make a prediction

```python
y_lr_train_pred=lr.predict(X_train)
y_lr_test_pred=lr.predict(X_test)

y_lr_train_pred
```

```python
y_lr_test_pred
```

#### Evaluate model performance (comparing the train and test set performance)

```python
from sklearn.metrics import mean_squared_error,r2_score
lr_train_mse = mean_squared_error(y_train,y_lr_train_pred)
lr_train_r2 = r2_score(y_train,y_lr_train_pred)

lr_test_mse = mean_squared_error(y_test,y_lr_test_pred)
lr_test_r2 = r2_score(y_test,y_lr_test_pred)
```

```python
print("LR mse (train):", lr_train_mse)
print("LR r2 (train):", lr_train_r2)

print("LR mse (test):", lr_test_mse)
print("LR r2 (test):", lr_test_r2)
```

```python
lr_results = pd.DataFrame(['Linear Regression',lr_train_mse,lr_train_r2,lr_test_mse,lr_test_r2]).transpose() #transpose means switch rows and columns
lr_results
```

```python
lr_results.columns= ['Method','Training MSE','Training R2','Test MSE','Test R2']
lr_results
```

### Random Forest

#### Training the model

```python
from sklearn.ensemble import RandomForestRegressor
rf = RandomForestRegressor(max_depth=2, random_state=100) # random state is for reproducibility means same result every time
rf.fit(X_train,y_train)

```

#### Applying the model to make a prediction

```python
y_rf_train_pred=rf.predict(X_train)
y_rf_test_pred=rf.predict(X_test)

y_rf_train_pred
```

```python
y_rf_test_pred
```

#### Evaluate the model performance

```python
from sklearn.metrics import mean_squared_error,r2_score
rf_train_mse = mean_squared_error(y_train,y_rf_train_pred)
rf_train_r2 = r2_score(y_train,y_lr_train_pred)

rf_test_mse = mean_squared_error(y_test,y_lr_test_pred)
rf_test_r2 = r2_score(y_test,y_lr_test_pred)
```

```python
print("RF mse (train):", lr_train_mse)
print("RF r2 (train):", lr_train_r2)

print("RF mse (test):", lr_test_mse)
print("RF r2 (test):", lr_test_r2)
```

```python
rf_results = pd.DataFrame(['RandomForest',rf_train_mse,rf_train_r2,rf_test_mse,rf_test_r2]).transpose() #transpose means switch rows and columns
rf_results
```

```python
rf_results.columns=['Method','Training MSE','Training R2','Test MSE','Test R2']
rf_results
```

### Model Comparison

```python
df_models = pd.concat([lr_results,rf_results] , axis = 0) #axis =0 row wise manner
df_models
```

```python
df_models.reset_index(drop=True)
```
